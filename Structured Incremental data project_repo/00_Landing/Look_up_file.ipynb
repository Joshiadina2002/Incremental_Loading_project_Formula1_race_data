**once Trigger Behavior:**
- **Processes all available data once and then stops**.
- If **no data is present at the time the query starts, the query will start and stop immediately**.
- If **you add new files after the query has stopped, the query will NOT restart automatically**.
- You must**manually start the query again to process new data.**
**Use Case:**

Useful for **batch-style processing**where you want to process all data currently available and then stop.

**2. availableNow Trigger Behavior:**

- Processes **all data that is available at the time the query starts and then stops**.
- If **no data is present, the query will start and stop immediately**.
- If **you add new files after the query has stopped, the query will NOT restart automatically.**
- You must **manually start the query again to process new data.**
**Use Case:**

**Designed for "micro-batch batch" processing,** especially in Databricks, to process all available data and then stop.
`Key Point:`
- **Neither once nor availableNow will automatically restart or detect new files after the query has stopped.**
- **If you want to process new files, you must start a new query with the same trigger.**

Summary Table
Trigger	Processes Existing Data?	Stops After?	Detects New Files After Stop?	Auto-Restarts?
once	Yes	All data	No	No
availableNow	Yes	All data	No	No
default	Yes	Never (until stopped)	Yes (while running)	N/A
Your Understanding
"When file placed again it will detect and start the availableNow query is it right?"

Not quite.

If you place a file after the query has stopped, it will NOT be processed automatically.
You must start a new query with availableNow to process the new file.
Typical Workflow with availableNow or once
Start the query with the trigger.
All available data is processed.
Query stops.
If new data arrives, start the query again to process it.
If You Want Continuous Processing:
Use the default trigger (no .trigger(...)) so the query keeps running and processes new files as they arrive.
This is not possible with availableNow or once triggers.
In summary:

Both once and availableNow process all available data and then stop.
Neither will process new files that arrive after the query has stopped.
You must manually restart the query to process new files.
