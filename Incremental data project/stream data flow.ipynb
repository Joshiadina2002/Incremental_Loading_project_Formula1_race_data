```python
dbutils.widgets.text("folder_name", "")
dbutils.widgets.text("itr", "")
```

```python
folder_name_list= list(dbutils.widgets.get('folder_name').split(','))
itr=dbutils.widgets.get('itr')
main_path="/Volumes/databricks_catalog/default/default_volume1"
```

```python
import time
d=dict()
for folder_name in folder_name_list:
    file_format=folder_name.split('@')
    print(file_format)
    d[file_format[0]]=file_format[1]
for f,ff in d.items():
    print(f"{f}:{ff}")
    dbutils.fs.cp(f"{main_path}/datasets/{f}/{f}_{itr}.{ff}", f"{main_path}/streaming_source/{f}/")
    print(f"{f}_{itr} file moved sucessfully from dataset to streaming source")
```
